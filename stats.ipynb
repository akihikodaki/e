## Initialization

```python
# SPDX-License-Identifier: Apache-2.0

from bokeh.io import output_notebook
from bokeh.plotting import figure, show
import json
import statistics

def ipc(core):
    return core['instructions'] / core['cycles']

def load(name):
    with open(f'stats/{name}.json') as io:
        return json.load(io)

output_notebook()

with open('benchmarks') as io:
    names = [line[:-1] for line in io]

stats = [load(name) for name in names]
ipcs = [ipc(phases[0]['roi']['cores'][0]) for phases in stats]
stats[0]
```

## IPC

### Per benchmark

```python
p = figure(y_range=names, height=2000)
p.hbar(y=names, right=ipcs)
p.x_range.start = 0
show(p)
```

### Mean

```python
statistics.mean(ipcs)
```
